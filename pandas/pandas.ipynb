# Selección de atributos - ordenar, seleccionar y modificar

```python
customers = pd.read_csv("Customers.csv");
customers.head(7)
```

```python
customers.describe()
```

```python
# buscar donde hay menos de 2 familiares, osea el family size <= 2
menos_de_dos = customers[customers["Family Size"] <= 2]
menos_de_dos
```

```python
# buscar donde hay menos de 2 familiares, osea el family size <= 2 y que el customer sea male
menos_de_dos = customers[(customers["Family Size"] <= 2) & (customers["Gender"] == "Male")]
menos_de_dos
```

# Datos específicos

```python
# mayores de 25 años pero menos de 75 años
customers[ (customers["Age"] > 25) & (customers["Age"] < 70) ]
```

```python
# conteo de profesiones
customers["Profession"].value_counts()
```

```python
customers[customers["Profession"] == "Engineer"]
```

# .loc búesqueda de filas por nombre

```python
# .loc accede a los valores de una fila por nombre
customers.loc[customers["Gender"] == "Male"]
```

```python
customers.loc[customers["Gender"] == "Female"]
```

```python
customers.loc[customers["Profession"] == "Doctor"]
```

# .iloc búesqueda de filas por indice

```python
# .iloc busca filas por posición
customers.iloc[3:7]
```

```python
# todos los valores menos el último
customers.iloc[:-1] # todos los valores menos el o los ultimos, -1, -2, -3, n numeros
```

```python
# si quiero los ultimos valores
customers.iloc[-1:]
```

```python
# si quiero los ultimos 5 valores
customers.iloc[-5:]
```
