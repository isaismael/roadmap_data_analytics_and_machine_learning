# KNN - K-Nearest Neighbrs - K-Vecinos más cercanos

- El algoritmo de K-NN es posiblemente el algoritmo de aprendizaje automatico mas sencillo.
- Para hacer una predicción de un nuevo punto de datos, el algoritmo busca los puntos de datos más cercanos en el conuunto de datos de entrenamiento: sus "vecinos mas cercanos"
- No tiene muchos parametros para configurar

# KNN - Clasificación
En su version más sencilla, se tiene un solo vecino más proximo, que es el punto de datos de entrenamiento más cercano al punto para el que queremos hacer una predicción. La predicción es entonces simplemente la salida conocida para este punto de entrenamiento.

Si agregamos más vecinos "K", se utiliza una botación para asignar una etiqueta. Esto significa que, para cada punto de prueba, contamos cuántos vecinos pertenecen a la clase 0 y cuántos a la clase 1. Posteriormente se le asigna la clase más cercana a cada vecino. es decir la clase más frecuiente entre los k vecinos más proximos



# KNN - Rregresión

- Usando un vecino cercano, la predicción es simplemente el valor objetivo del vecino más cercano.
- Cuando se utiliza varios vecinos más próximos, la predicción es la media de los vecinos pertinentes.
- En vez de un voto seria un promedio.
