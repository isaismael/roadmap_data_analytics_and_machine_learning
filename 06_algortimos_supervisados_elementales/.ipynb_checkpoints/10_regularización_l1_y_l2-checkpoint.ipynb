# Ridge Regression y Lasso Regression

### Ridge Regression 
- Regularización significa restringir explicitamente un modelo para evitar el sobreajuste.
- El tipo particular utilizado por la regresión de cresta se conoce como regularización **L2**
- Busca que todas las entradas de W deben ser cercanas a cero. Intuititvamente, esto significa que cada característica debe tener el menor efecto posible sobre el resultado (lo qu se traduce en tener una pedinete pequeña), sin dejar de predecir bien.

# Lasso Regression
- El uso del Lasso también restringe los coeficientes para que sean prpoximos a cero, pero de una forma ligeramente diferente, denominada regularización **L1**.
- La consecuencia de la regularización **L1** es que cuando se utiliza el lasso, algunos coeficientes son exactamente cero. Esto significa que algnas características son totalmente ignoradas por el modelo.
- Esto puede verso como una forma de selección automática de características a menudo hace que un modelo sea mpás de interpretar y puede revelar caracteristicas más importante de su modelo

# En otras palabras
Ambos lo que hacen es quitar caracteristicas que no tan importantes para la predicción de nuestro modelo.

Si tenemos un dataset con una columna con datos repetitivos, estos 2 tipos de regularización es discriminar esas caracteristicas, estandarizar el sistema y hacer que nuestra salida en consideración valores que realmente sean importante para nuestra salida.
