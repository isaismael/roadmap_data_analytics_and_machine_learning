# Regresión Logistica

- Se suele aplicar para calcular la probabilidad de que una instancia pertenezca a una clase determinada, si una propabilidad es superior al 50% puede pertenecer a una clase, en el caso de una clasificación binaria.
- Tambien se uede aplicar para clasificación múltiple.
- **IMPORTANTE:** A pesar de si nombre este algoritmo se usa para **CLASIFICAR**
