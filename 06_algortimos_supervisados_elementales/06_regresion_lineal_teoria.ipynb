# Linear Regression

## Minimos cuadrados ordinarios (MCO)

La regresión lineal es el método lineal más sencillo y clásico para la regresión, encuentra los parámentros w y b 
que minimizan el error cuadrático medio entre las predicciones y valores verdaderos del conjunto de entrenamiento.

y = [w].[x] + [b]

La regresión lineal no tiene parámetros, lo que es una ventaja, pero tampoco tiene forma de controlar la complejidad del modelo.

- W equivale a la pendiente
- B pertenece a las intersecciones que podriamos tener en el modelo
- X es el parametros de entrada.

Nosotros vamos a tener un conjunto de parametros mas amplio.

W -> seria como un sistema matricial, seria como un conjunto de datos.

Y -> Es el valor de salida

Regresion lineal es muy importante porque tiene como un inicio a redes neuronales.

## Ejemplo en codigo

-> from sklearn.linear_model import LinearRegression

-> lin_reg = LinearRegression()

-> lin_reg.fit(x, y)

-> lin_reg.intercept_, lin_reg.coef_

Para una cantidad grande valores el modelo se vuelve lento en entrenamiento, po lo que existen ciertas alternativas.
