## Gradient Descent

- El Descenso de Gradiente es un algorirmo de optimización muy genérico capaz de encontrar soluciones óptimas para una amplia gama de problemas
- La idea general del Descenso de Gradiente es ajustar los parámetros de forma iterativa para minimizar una función de costo.
- Mide el gradiente local de la función de error con respecto al vector de parámetros, y va en la dirección del gradiente descente, dando un pequeño paso a la vez intentando disminuir la función de costo. Una vez que el gradiente es cero ¡Has alcanzado un mínimo!

E-learning step, son pequeños pasos para que el modelo vaya aprendiendo

Un parámetro importante a considerar es el numero de pasos determinado por el *learning rate*

Si el learning rate es muy pequeño, el algortimo necesitara más iteraciones para converger.

Si por otro lado el learning rate es muy grande, podriá saltar a través del valle haciendo que el algoritmo nunca llegue al mínimo
