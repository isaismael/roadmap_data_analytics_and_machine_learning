# Regresión Logistica

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
# metrics
from sklearn.metrics import confusion_matrix
from sklearn.metrics import ConfusionMatrixDisplay
from sklearn.metrics import classification_report
```

```python
# solver
# para pequeños dataset 'liblinear' es una buena opcione
# pero para problemas multiclases solo podemos usar ->  ‘lbfgs’, ‘newton-cg’, ‘newton-cholesky’ and ‘sag’}
# hay que elegir la penalización
```

```python
x_train, x_test, y_train, y_test = train_test_split(df_mobile.drop(columns="price_range"), df_mobile.price_range, train_size=0.8, random_state=123)
```

```python
# el random state hay que poner un valor aleatorio
log_reg = LogisticRegression(l1_ratio=0, solver="sag", random_state=42, max_iter=10000)
log_reg
```

```python
log_reg.fit(x_train, y_train)
```

```python
y_pred = log_reg.predict(x_test)
```

```python
log_reg.score(x_test, y_test)
```

```python
print(classification_report(y_test, y_pred, zero_division=0))
```

```python
cm = confusion_matrix(
    y_true= y_test,
    y_pred=y_pred
)
```

```python
disp = ConfusionMatrixDisplay(cm)
disp.plot()
plt.show()
```
