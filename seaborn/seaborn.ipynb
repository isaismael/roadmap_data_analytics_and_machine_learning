```python
import pandas as pd
import seaborn as sns
```

```python
# enconding es cuando tenemos caracteres especiales como ñ
df_ph = pd.read_csv("lettuce_dataset.csv", encoding="latin1")
df_ph.head(5)
```

# Gráficas de barra

```python
# primer argumento es el dataframe
sns.barplot(data=df_ph, x="pH Level", y="Plant_ID", errorbar=None)
```

# Barra de distribuccion - scatter

```python
sns.scatterplot(data=df_ph, x="TDS Value (ppm)", y="Temperature (°C)")
```

# Gráfica de lineas

```python
sns.lineplot(data=df_ph, x="Date", y="pH Level", errorbar=None)
```

# Gráficas básicas

### Histograma

```python
sns.histplot(data=df_ph, x="TDS Value (ppm)")
```

### Gráfica linea al estilo gauss - KDE PLOT

```python
sns.kdeplot(data=df_ph, x="Temperature (°C)")
```

### Gráfica de cajas

```python
sns.boxplot(data=df_ph, x="Humidity (%)", y="Growth Days")
```

### Gráfica mas recomendada

```python
# la grafica de todas nuestras variables numeras, solo las que tienen numeros
sns.pairplot(data=df_ph)
```

### Paleta de colores y hue

```python
## hue colorea de acuerdo al calor 
# es dar color en base a una caracteristica en especial
sns.scatterplot(data=df_ph, x="TDS Value (ppm)", y="Temperature (F)", hue="pH Level", palette="colorblind")
```
