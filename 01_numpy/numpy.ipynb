## Funciones universales
Son operaciones 1 a 1, suma, resta, mul, div, potenciasion, etc

```python
array1 = np.array([8,13,17])
array2 = np.array([25,15,5])
```

```python
# Suma
np.add(array1, array2)
```

```python
# Resta - tiene order jerarquico
np.subtract(array1, array2)
```

```python
# Multiplicacion - tmb jerarquico, osea indice 0 con indice 0 y asi...
np.multiply(array1, array2)
```

```python
np.power(array1, np.array([2,3,2]))
```

```python
# Raiz cuadrada
np.sqrt(array1)
```

```python
# El cuadrado del valor, osea elevar al cuadrado
np.square(array2)
```

```python
# Máximo común divisor, tiene 2 argumentos
np.gcd(array1, array2)
```

```python
# Mínimo común multiplo
np.lcm(array1, array2)
```

```python
# Valor absoluto - quitar o volver positivo un valor en un array
x1 = np.array([-3, -5, -4, -2])
np.abs(x1)
```

## Funciones trigonometricas y logaritmicas.

```python
# antes de de ver estar funciones veremos linspace, es un array pre-producido,
# que lo que hace es poner un valor inicial, un valor final y cuantos valores se quiere entre estos 2 extremos
```

```python
# 2 argumentos, valor inicial desde donde lo quiere hasta donde lo quiero, y los intervalor
x2 = np.linspace(0, np.pi, 3)
x2
```

```python
print("ánguo original:", x2)
print("Seno del ángulo:", np.sin(x2))
print("Coseno del ángulo:", np.cos(x2))
print("Tangente del ángulo:", np.tan(x2))
```

### Funciones exponenciales

```python
x3 = np.array([1, 2, 3])
print("x:", x3)
print("e x:", np.exp(x3))
print("2 x:", np.exp2(x3))
print("3 x:", np.power(3, x3))
```

### Funciones logaritmicas

```python
# logaritmo natural de x
print("ln(x):", np.log(x3))
# logaritmo base 2
print("log2:", np.log2(x3))
# logaritmo base 10
print("log10:", np.log10(x3))
```

# Comparación de arrays

```python
# x1 > x2
np.greater(array1, array2)
```

```python
# x1 >= x2
np.greater_equal(array1, array2)
```

```python
# x1 < x2
np.less(array1, array2)
np.less_equal(array1, array2)
```

```python
# x1 == x2 
np.equal(array1, array2)
```

```python
# x1 != x2
np.not_equal(array1, array2)
```

# funciones estadisticas

```python
# arma array random, parametros que le pasamos la cantidad de filas y segundo colmuna
np.random.rand(2, 5)
```

```python
# randint, los valores a mostrar serán enteros 
# el primero valor denota el limite inferior que quiero, y el segundo el limite superior
# el size es la calidad de valores que vas a querer de ese rango
np.random.randint(8,100, size=10)
```

# Máximo y mínimo

```python
v1 = np.random.randint(0,50, size=15)
v1
```

```python
# armar una matriz con random
v2 = np.random.randint(0,50, size=(3, 3))
v2
```

```python
# minimo 
print("El min de un array es: ", np.amin(v1))
print("El min de una matriz es: ", np.amin(v2))
```

```python
# máximo
print("El max de un array es: ", np.amax(v1))
print("El max de una matriz es: ", np.amax(v2))
```

# Funciones estadisticas de uso muy común
### Percentiles, media, mediana, varianza, y desviación estándar

```python
# Percentil, es el núm. que se encuentra a una cierta distancia entre el 0 y el 100
print("percentil del array 32:", np.percentile(v1,32))
print("percentil de la matriz 25:", np.percentile(v2,25))
```

```python
# mediana, es el valor que se encuentra en medio
print("Mediana:", np.median(v1))
# no confundir con la media, es el promedio de todos los valores, la suma de los valores divido entre el total de los valores
print("Media:", np.mean(v1))
```

```python
# varianza y desviacion estandar
# es cuanto se aleja del promedio, y como funciona esa distribucion de valores
print("Desviación standar:", np.std(v1))
# la varianza es un numero mas amplio, debido a su naturaleza de la raiz cuadrada
print("Varianza:", np.var(v1))
```
