# Estética de gráficas

plt.plot(x,y, Estilo de line, color, ancho de trazo)

Estilo de línea: '-' or 'solid': solid line
'--' or 'dashbed': dashbed line
'-.' or 'dashdot': dash-dotted-line
':' or 'dotted': dotted line
'none', 'None', ',or': draw nothing

Color: 'b' as blue
'g' as green
'r' as red
'c' as cyan
'm' as magenta
'y' as yellow
'k' as black
'w' as white

Ancho: linewidth=valor

```python
x = np.linspace(0, 10, 100)
plt.plot(x, np.sin(x), ':', color='r', linewidth=3)
plt.xlabel("Radianes")
plt.ylabel("Valores")
plt.suptitle("Función senoidal")
plt.title("Ejemplo de matplotlib.pyplot")
plt.show()
```

# Grid y Axis
hacer una grilla o rejilla


```python
x = np.linspace(0, 10, 100)
plt.plot(x, np.sin(x), ':', color='r', linewidth=3)
plt.xlabel("Radianes")
plt.ylabel("Valores")
plt.suptitle("Función senoidal")
plt.title("Ejemplo de matplotlib.pyplot")
# grid
plt.grid()
# que pasa si nuestro grafico no se ajusta a la medida que buscamos
# los primeros 2 valores que se le pasan son de x
# los ultimos 2 valores que se le pasan son de y
# osea el tamaño de la cuadricula
plt.axis([-1, 11, -1.5,1.5])
plt.show()
```

```python
# impresion de distribuciónm, esto es solo un ejemplo
rng = np.random.RandomState(0)
# randn, n para que sean naturales
x = rng.randn(100)
y = rng.randn(100)
colors = rng.rand(100)
sizes = 1000 * rng.rand(100)
# cmap="viridis es la paleta de colores
plt.scatter(x,y, c=colors, s=sizes, alpha=.3, cmap="viridis")
# color var nos inidca los colores que estamos usando
plt.colorbar()
plt.show()
# esto es solo un ejemplo mas para ver lo visual, los datos no tienen relación alguna
```

# Aplicación: Procesando Datos

```python
clima = pd.read_csv("GlobalWeatherRepository.csv")
clima.head()
```

```python
# agrupar por country y calcular el promedio de temperaturas
df_avg_temp = clima.groupby("country", as_index=False)['temperature_celsius'].mean()
df_avg_temp.head(9)
```

```python
df_avg_temp.info()
```

```python
# tomamos la temperaturas mas bajas
min_temp = df_avg_temp.nsmallest(15, "temperature_celsius").reset_index()
```

```python
min_temp.head(15)
```

```python
# eliminar el indice
min_temp = min_temp.drop(columns="index")
min_temp
```

```python
# tamño de la figura, argumentos son el ancho y alto - son pulgadas
plt.figure(figsize=(10, 6))
plt.plot(min_temp["country"], min_temp["temperature_celsius"], color="m", linestyle="dotted", label="Paises con temps. bajas")
# legend() es para traer nuestro label interno, osea info de que se trata la gráfica
plt.legend()
plt.show()
```

# Ordenamos un poco mejor la gráfica

## Rotación de rótulos en "x"

```python
plt.figure(figsize=(10, 6))
plt.plot(min_temp["country"], min_temp["temperature_celsius"], color="m", linestyle="dotted", label="Paises con temps. bajas")
plt.legend()
plt.title("Analisis de temperaturas bajas")
plt.xlabel("Paises")
plt.ylabel("Temperatura (C)")
# tick_params hace que sea un poco más legible los lables en el eje x
plt.tick_params(axis="x", labelrotation=60)
# otros parametros que se pueden modificar
plt.tick_params(direction="out", labelsize="small")
# lineas arriba de los labels
plt.tick_params(color="red")
plt.show()
```

# Anotaciones

```python
plt.figure(figsize=(10, 6))
plt.plot(min_temp["country"], min_temp["temperature_celsius"], color="m", linestyle="dotted", label="Paises con temps. bajas")
plt.legend()
plt.title("Analisis de temperaturas bajas")
plt.xlabel("Paises")
plt.ylabel("Temperatura (C)")
plt.tick_params(axis="x", labelrotation=60, direction="out", labelsize="small", color="red")
# annotate, lleva 8 parametros
# los primeros 4 comandos, son el nombre de anotacion, la posición en la que queremos que se encuentre, el tamaño de la letra y el color
# los siguientes 4 comando, son con la flecha o linea que querramos coloras
plt.annotate("Argentina", xy=(13, 13.5), fontsize=14, color="green", xytext=(12.35, 14.5), arrowprops=dict(arrowstyle="-|>", color="red"))
plt.show()
```

# Líneas horizontales y verticales

```python
plt.figure(figsize=(10, 6))
plt.plot(min_temp["country"], min_temp["temperature_celsius"], color="m", linestyle="dotted", label="Paises con temps. bajas")
plt.legend()
plt.title("Analisis de temperaturas bajas")
plt.xlabel("Paises")
plt.ylabel("Temperatura (C)")
plt.tick_params(axis="x", labelrotation=60, direction="out", labelsize="small", color="red")
plt.annotate("Argentina", xy=(13, 13.5), fontsize=14, color="green", xytext=(12.35, 14.5), arrowprops=dict(arrowstyle="-|>", color="red"))
# axhline - horizontal, axvline - vertical
plt.axhline(y=10, xmin=0, xmax=1, linewidth=2, linestyle='--', color="gray")
# agregamos una anotacion para que se entienda a que se refiere la linea
plt.annotate("Temperaturas \n menos a 10 °C", xy=(0.5, 10.1), fontsize=12, ha="left")
plt.show()
```
