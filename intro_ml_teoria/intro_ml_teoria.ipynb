# Regresión
## Métricas de evalución
Para medir la precisón de nuestro modelo es necesario someterlo a alguna prueba de error:

#### Error cuadrático medio, (mean squarre error, MSE): 
La media de la diferencia entre el valor real y el valor predicho o estiamdo al cuadrado


#### Error Absoluto medio, (mean absolute error, MAE):
Se define como la difrencia en valor absoluto entre el valor real y el valor predicho


#### Raíz del error cuadrático medio, (root mean square error, RMSE):
Se define como la raíz cuadrada de la media de la difrencia entre el valor real y el valor predicho al cuadrado, en comparación con MAE amplifica y penaliza los errores grandes.



#### coeficiente de determinación R2
Mide lo bien que un modelo de regresión se ajusta a los datos reales. En otras palabras, se trata de una medida de la precisión general del modelo. R al cuadrado es también conocido como el coeficiente de determinación.
R2 se mide en escala de 0 a 1. Un valor de 1 indica un modelo que predice perfectamente los valores en el de destino. Un valor de 0 indica un modelo qu eno tiene ningun valor predictivo

# Métricas de clasificación generales
### Métricas comúnes
- Accuracy: Representa la proporción del número de predicciónes correctas entre el número total de predicciones.


- Precisión: Indica la proporción de ejemplos que son verdaderamente positivos


- Recall: Indica que tan completos son los resultados. La métrica recall equivale a la métrica de sensibilidad. Un gran call captura un gran porentaje de ejemplos positivos.


- F-measure(f1): Es una métrica que combina preción y call utilizando la media armónica, Simplifica el rendimiento de un algoritmo de clasificación a una única métrica



## Posibilidades de resultados
- Alta precision y alto recall: El modelo maneja perfectamente esa clase.
- Alta precision y bajo reacall: El model no detecta la clase muy bien, pero cuando lo hace es altamente confiable.
- Baja precisión y alto recall: La clase detecta bien la clase pero también incluye muestras de otras clases.
- Baja precisión y bajo recall: El modelo no logra clasificar la clase correctamente

## Curvas ROC, y AUC (area bajo la curva)
- La curva roca se sentra en la diagonal, en cuanto mas nos acercamos el valor extremo izquierdo, mejor será nuestra predicción
- En el caso de clasificacion binaria, se puede utuliza una métrica obtenida a partir de las curvas receiver operating characteristi (ROC)
- Esta métrica se utiliza para determinar el balanca entre la detección de verdaderos positivos y evitar falsos positivos. Para ello, se muestra la proporción de detección de los verdaderos positivos en el eje vertical y la proporción de los falsos positivos en el eje horizontal, para un umbral o punto de corte determinado.
- A medida que al curva este más cercana de la esquina superior izquierda, mejor será.
- A partir de ROC se puede obtener area under the curve (AUC) o en castillano area bjo la curva.
- Esta métrica va desde valroes de 0,5 para un clasificador sin potencia predictiva (completamente aleatorio) hasta 1 para un clasificador perfecto.
