# Aplicaciones de machine learning
### Aprendizaje Supervisado
- Clasificación (Datos categóricos) - datos con cualidades, alto, mediano, bajo, colores, precios (alto, medio, bajo), tambien puede ser con imagenes
- Regresión (Datos Numéricos) - trata de valores numericos, osea predecir un valor, un suceso que aun no pasa.
- Algunos algoritmos de Aprendizaje supervisado:
    - K-Nearest Neighbors (K-vecinos cercanos)
    - Linear Regression (Regresión lineal)
    - Logistic Regression (Regresión logistica)
    - Support Vector Machines - SVMs (Maquinas de vectores de soporte)
    - Desición Trees and Random Forest (Arboles de decisión)
    - Neural Networks (Redes Neuronales)

### Aprendizaje No Supervisado
- Clustering (Agrupación)
- Detección de anomalias
- Algunos algoritmos de Aprendizaje no supervisado:
    - K-Means (Las K-Medias)
    - DBSCAN
    - Hierarchical Cluster Analysis (HCA)
    - Anomaly detection and novelty detection (Maquinas de vectores de soporte)
    - One-class SVM
    - Isolation Forest

### Aprendizaje por refuerzo
- Agentes inteligentes (El aprendizaje por reforzamiento se basa principalmente por agentes inteligentes)

### Tenemos 3 tipos de aproximaciones: Regression, Clasificacion y Deep Learning
- La regresión busca predecir valores númericos, sea enteros, decimales, no importa, pero si aproxima una numeracion de prediccion futura
- La clasificacion en base a un conjunto de información pueda dicernir si pertenece a un conjunto o a otro, o a uno de varios conjunto, osea darle una etiqueta a ese nuevo conocimientos que está entrando
- El DP resuelve regresión y clasificacion pero con ANN




### Underfittig: Entreno al modelo con 1 sola raza de perror. La maquina fallará en reconocer al perro por falta de suficientes muestras. No puede generalizar el conociminetos
### Overfiting: Entreno al modelo con 10 razas de perros marrones. La maquina fallará en reconocer un perro nuevo porque no tiene estrictamente los mismos valores de las muestras de entrenamiento.

# Regresión
## Métricas de evalución
Para medir la precisón de nuestro modelo es necesario someterlo a alguna prueba de error:

#### Error cuadrático medio, (mean squarre error, MSE): 
La media de la diferencia entre el valor real y el valor predicho o estiamdo al cuadrado


#### Error Absoluto medio, (mean absolute error, MAE):
Se define como la difrencia en valor absoluto entre el valor real y el valor predicho


#### Raíz del error cuadrático medio, (root mean square error, RMSE):
Se define como la raíz cuadrada de la media de la difrencia entre el valor real y el valor predicho al cuadrado, en comparación con MAE amplifica y penaliza los errores grandes.
